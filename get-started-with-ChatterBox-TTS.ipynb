## ⚡ Step 2: Chatterbox-Turbo - The Fastest & Most Fun Model!

**Chatterbox-Turbo** is the newest addition to the family with a streamlined **350M parameter** architecture. What makes it special?

### 🎭 Paralinguistic Tags (SUPER COOL!)
You can add **natural expressions** directly in your text:
- `[laugh]` - Full laugh
- `[chuckle]` - Light chuckle  
- `[cough]` - Natural cough
- `[sigh]` - Sighing sound
- And more!

Let's see it in action!

```python
from huggingface_hub import login
from google.colab import userdata

HF_TOKEN = userdata.get('HF_TOKEN')
login(token=HF_TOKEN)
```

```python
import torchaudio as ta
from chatterbox.tts_turbo import ChatterboxTurboTTS
from IPython.display import Audio, display

# Load the Turbo model (this may take a moment to download ~1.5GB)
print("Loading Chatterbox-Turbo model...")
turbo_model = ChatterboxTurboTTS.from_pretrained(device="cpu")
print("Model loaded successfully!")
```

```python

# 🎉 Example 1: Customer Service Agent with a Chuckle
text1 = """Oh, that's hilarious! [chuckle] Um anyway, we do have a new model in store.
It's the SkyNet T-800 series and it's got basically everything.
Including AI integration with ChatGPT and all that jazz.
Would you like me to get some prices for you?"""

print("\n🎙️ Generating: Customer Service Agent")
wav1 = turbo_model.generate(text1)
ta.save("turbo_demo1.wav", wav1, turbo_model.sr)
display(Audio("turbo_demo1.wav"))
```

```python
# 🎭 Example 2: Storyteller with Multiple Expressions
text2 = """And then, [laugh] you won't believe what happened next!
The dragon [cough] excuse me, the dragon just sat down and asked for tea!
[chuckle] Can you imagine? A fire-breathing dragon, wanting Earl Grey!"""

print("🎙️ Generating: Fantasy Storyteller")
wav2 = turbo_model.generate(text2)
ta.save("turbo_demo2.wav", wav2, turbo_model.sr)
display(Audio("turbo_demo2.wav"))
```

```python
# 🎬 Example 3: Movie Trailer Style
text3 = """In a world where AI has taken over... [cough]
One programmer must rise against the machines.
This summer... [laugh] just kidding, the AI are actually pretty friendly!"""

print("\n🎙️ Generating: Movie Trailer Parody")
wav3 = turbo_model.generate(text3)
ta.save("turbo_demo3.wav", wav3, turbo_model.sr)
display(Audio("turbo_demo3.wav"))
```

## 🎛️ Step 3: Original Chatterbox - Fine-Grained Control

The original **Chatterbox** model offers powerful controls for tweaking your speech:

### Key Parameters:
- **`exaggeration`** (0.25 - 2.0): Controls emotional intensity. Default 0.5 is neutral.
- **`cfg_weight`** (0.0 - 1.0): Controls pacing/tempo. Lower = faster, higher = slower.
- **`temperature`**: Controls randomness (default 0.8)

### Pro Tips:
- For **dramatic speech**: Low `cfg_weight` (~0.3) + High `exaggeration` (~0.7)
- For **calm narration**: Default settings work great
- For **fast speakers**: Lower `cfg_weight` to ~0.3

```python
from chatterbox.tts import ChatterboxTTS

# Load the original Chatterbox model
print("Loading original Chatterbox model...")
model = ChatterboxTTS.from_pretrained(device="cpu")
print("Model loaded!")

# The text we'll use for all variations
epic_text = """Now let's make my mum's favourite. So three mars bars into the pan.
Then we add the tuna and just stir for a bit, just let the chocolate and fish infuse.
A sprinkle of olive oil and some tomato ketchup. Now smell that.
Oh boy this is going to be incredible!"""
```

```python
# 🎯 Example 1: Default Settings (Balanced)
wav_default = model.generate(
    epic_text,
        exaggeration=0.5,
            cfg_weight=0.5,
                temperature=0.8
                )
ta.save("chatterbox_default.wav", wav_default, model.sr)
display(Audio("chatterbox_default.wav"))
```

```python
# 🔥 Example 2: DRAMATIC (High Exaggeration - More dramatic!)
wav_dramatic = model.generate(
    epic_text,
        exaggeration=0.9,  # Cranked up!
            cfg_weight=0.3,     # Faster pacing
                temperature=0.8
    )
ta.save("chatterbox_dramatic.wav", wav_dramatic, model.sr)
display(Audio("chatterbox_dramatic.wav"))
```

```python
# 😌 Example 3: CALM (Low Exaggeration, Higher CFG, Calm and Measured)
wav_calm = model.generate(
        epic_text,
            exaggeration=0.3,   # More subdued
                cfg_weight=0.7,     # Slower, more deliberate
                    temperature=0.6     # Less random
                    )
ta.save("chatterbox_calm.wav", wav_calm, model.sr)
display(Audio("chatterbox_calm.wav"))
```
