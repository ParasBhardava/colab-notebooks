
## 2️⃣ Custom Voice Generation

Now let's use the **CustomVoice model** to generate speech using Qwen3-TTS's built-in premium voices. This model supports 9 carefully crafted voices across multiple languages!

### 🎭 Available Speakers

| Speaker | Description | Native Language |
|---------|-------------|-----------------|
| **Vivian** | Bright, slightly edgy young female | Chinese |
| **Serena** | Warm, gentle young female | Chinese |
| **Uncle_Fu** | Seasoned male with low, mellow timbre | Chinese |
| **Dylan** | Youthful Beijing male, clear & natural | Chinese (Beijing) |
| **Eric** | Lively Chengdu male, slightly husky | Chinese (Sichuan) |
| **Ryan** | Dynamic male with strong rhythm | English |
| **Aiden** | Sunny American male, clear midrange | English |
| **Ono_Anna** | Playful Japanese female, light & nimble | Japanese |
| **Sohee** | Warm Korean female with rich emotion | Korean |

The best part? You can control emotions and speaking style with natural language instructions!

```python
# 🎤 Custom Voice Generation with Emotion Control
import torch
import soundfile as sf
from qwen_tts import Qwen3TTSModel
from IPython.display import Audio, display

# Load the CustomVoice model (using the 0.6B version for faster loading)
print("📥 Loading Qwen3-TTS CustomVoice model...")
custom_voice_model = Qwen3TTSModel.from_pretrained(
    "Qwen/Qwen3-TTS-12Hz-0.6B-CustomVoice",
    device_map="cuda:0",
    dtype=torch.bfloat16,
)
print("✅ Model loaded successfully!")

# Check available speakers and languages
print("\n🎭 Available Speakers:", custom_voice_model.get_supported_speakers())
print("🌍 Supported Languages:", custom_voice_model.get_supported_languages())
```

```python
# 🎭 Generate Speech with Different Emotions!
# Let's try the same text with different emotional instructions

text = "I can't believe this is actually happening right now!"

# Generate with different emotions
emotions = [
    ("neutral", "Speak in a calm, neutral tone."),
    ("happy", "Speak with excitement and joy!"),
    ("angry", "Speak with frustration and anger."),
    ("sad", "Speak with a melancholic, disappointed tone.")
]

print("🎬 Generating speech with different emotions...\n")

for emotion_name, instruct in emotions:
    wavs, sr = custom_voice_model.generate_custom_voice(
        text=text,
        language="English",
        speaker="Ryan",
        instruct=instruct,
    )

    # Save and play
    filename = f"emotion_{emotion_name}.wav"
    sf.write(filename, wavs[0], sr)
    print(f"🎧 {emotion_name.upper()}: \"{instruct}\"")
    display(Audio(wavs[0], rate=sr))
    print()
```


## 3️⃣ Voice Design - Create Custom Voices from Descriptions!

This is where Qwen3-TTS really shines! 🌟

With **Voice Design**, you can create entirely new voices just by describing them in natural language. Want a gruff pirate voice? A cheerful child? An elderly storyteller? Just describe it!

> 🎨 **Voice Design** lets you craft unique character voices without needing any audio samples - perfect for game development, audiobook production, and creative projects!

```python
# 🎨 Voice Design - Create Voices from Descriptions
# Load the VoiceDesign model
print("📥 Loading Qwen3-TTS VoiceDesign model...")
voice_design_model = Qwen3TTSModel.from_pretrained(
    "Qwen/Qwen3-TTS-12Hz-1.7B-VoiceDesign",
    device_map="cuda:0",
    dtype=torch.bfloat16,
)
print("✅ VoiceDesign model loaded!")

# Let's create some unique character voices!
characters = [
    {
        "name": "Wise Old Wizard",
        "text": "Young one, the path to wisdom is paved with patience and perseverance.",
        "instruct": "Elderly male voice, deep and gravelly, speaking slowly with wisdom and authority. British accent, as if from an old fantasy tale.",
        "language": "English"
    },
    {
        "name": "Energetic Anime Girl",
        "text": "すごい！これは本当に面白いですね！",
        "instruct": "Young female voice, high-pitched and energetic, with an enthusiastic and playful tone typical of anime characters.",
        "language": "Japanese"
    },
    {
        "name": "News Anchor",
        "text": "Breaking news tonight: scientists have made a groundbreaking discovery.",
        "instruct": "Professional male voice, clear and authoritative, with perfect enunciation like a prime-time news anchor.",
        "language": "English"
    }
]

print("\n🎭 Creating custom character voices...\n")

for char in characters:
    print(f"🎬 Generating: {char['name']}")
    print(f"   Voice description: {char['instruct'][:60]}...")

    wavs, sr = voice_design_model.generate_voice_design(
        text=char['text'],
        language=char['language'],
        instruct=char['instruct'],
    )

    filename = f"voice_design_{char['name'].replace(' ', '_').lower()}.wav"
    sf.write(filename, wavs[0], sr)
    display(Audio(wavs[0], rate=sr))
    print()
```


## 4️⃣ Voice Cloning - Clone Any Voice with Just 3 Seconds!

One of the most impressive features of Qwen3-TTS is its ability to clone voices with incredible accuracy using just **3 seconds of reference audio**!

### How it Works:
1. Provide a short audio sample of the voice you want to clone
2. Include a transcript of what's being said in the sample
3. Generate new speech that sounds like the cloned voice

> ⚠️ **Important:** Always get permission before cloning someone's voice! Voice cloning should be used responsibly and ethically.

```python
# 👥 Voice Cloning - Clone Voices from Audio Samples
# Load the Base model for voice cloning
print("📥 Loading Qwen3-TTS Base model for voice cloning...")
clone_model = Qwen3TTSModel.from_pretrained(
    "Qwen/Qwen3-TTS-12Hz-0.6B-Base",
    device_map="cuda:0",
    dtype=torch.bfloat16,
)
print("✅ Clone model loaded!")

# Use a sample audio URL provided by Qwen team
# Attempt to download the reference audio locally due to HTTP 403 Forbidden error
ref_audio_url = "clone.wav" # Use the locally downloaded file
ref_text = """ Now this is using nano banana to create images based on the text prompt, but the best part is that it has native integrations. """

print("\n🎤 Reference audio transcript:", ref_text[:50] + "...")

# Generate new content with the cloned voice
new_texts = [
    "Welcome to our channel! Today we're exploring the amazing world of AI voice synthesis.",
    "The weather is absolutely beautiful today. Perfect for a walk in the park!",
    "I can't believe how realistic this voice cloning technology has become."
]

print("\n🔊 Generating speech with cloned voice...\n")

for i, text in enumerate(new_texts, 1):
    wavs, sr = clone_model.generate_voice_clone(
        text=text,
        language="English",
        ref_audio=ref_audio_url,
        ref_text=ref_text,
    )

    filename = f"voice_clone_sample_{i}.wav"
    sf.write(filename, wavs[0], sr)
    print(f"📝 Sample {i}: \"{text[:50]}...\"")
    display(Audio(wavs[0], rate=sr))
    print()
```


## 5️⃣ Advanced: Design a Voice, Then Clone It!

This is the **ultimate workflow** for creating consistent character voices! 🎬

### The Power Combo:
1. **Design** a unique voice using natural language descriptions
2. **Save** that voice as a reference
3. **Clone** it to generate unlimited consistent dialogue

This is perfect for:
- 🎮 Game character dialogue
- 📚 Audiobook narration with consistent characters  
- 🎬 Animation voice acting
- 🎙️ Podcast hosts and recurring characters

```python
# 🎯 Advanced Workflow: Design Voice → Clone for Consistency
# Perfect for creating persistent character voices!

# Step 1: Design your character's voice
print("🎨 Step 1: Designing a unique character voice...")

character_voice_description = """
Male, 17 years old, tenor range, gaining confidence - deeper breath support now,
though vowels still tighten when nervous. American high school student voice.
"""

# Create a reference audio with the designed voice
ref_text = "H-hey! You dropped your... uh... calculus notebook? I mean, I think it's yours? Maybe?"

ref_wavs, sr = voice_design_model.generate_voice_design(
    text=ref_text,
    language="English",
    instruct=character_voice_description,
)
sf.write("character_reference.wav", ref_wavs[0], sr)
print("✅ Character voice designed!")
display(Audio(ref_wavs[0], rate=sr))

# Step 2: Create a reusable clone prompt
print("\n🔧 Step 2: Creating reusable voice prompt...")
voice_clone_prompt = clone_model.create_voice_clone_prompt(
    ref_audio=(ref_wavs[0], sr),
    ref_text=ref_text,
)
print("✅ Voice prompt created!")

# Step 3: Generate multiple lines with consistent voice
print("\n🎬 Step 3: Generating dialogue with consistent character voice...\n")

dialogue_lines = [
    "No problem! I actually... kinda finished those already? If you want to compare answers...",
    "What? No! I mean yes but not like... I just think you're... your technique is really precise!",
    "See you tomorrow! I mean, if you want to. No pressure. Totally cool either way!",
]

for i, line in enumerate(dialogue_lines, 1):
    wavs, sr = clone_model.generate_voice_clone(
        text=line,
        language="English",
        voice_clone_prompt=voice_clone_prompt,  # Reuse the same voice!
    )
    sf.write(f"character_dialogue_{i}.wav", wavs[0], sr)
    print(f"🗣️ Line {i}: \"{line[:50]}...\"")
    display(Audio(wavs[0], rate=sr))
```
